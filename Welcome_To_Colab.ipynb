```
import pandas as pd

# Load movie data
df = pd.read_csv("/content/IMDB Dataset (1).csv")

# Get all genres
genres = sorted(set(
     genre
     for movie_genres in df["Genre"].dropna()
     for genre in movie_genres.split(", ")
    ))

# Get genre from user
def get_genre():
    print("\n Available Genres: ")

# Print
    for i, genre in enumerate(genres, 1):
        print(i, genre)

    while True:
        choice = int(input("\n Enter Genre Number: "))
        if 1 <= choice <= len(genres):
            return genres[choice - 1]  # Start From 0
        else:
            print("Invalid choice. Please try again.")

# Recommend movies
def recommend(genre):
    # get all the movies based on the genres
    movies = df[df["Genre"].str.contains(genre, case=False, na=False)]

    if movies.empty:
        print("No movies found.")
        return

    movies = movies.sample(min(5, len(movies)))

    print("\n Movie Recommendations:")
    for _ , movie in movies.iterrows():
        print(movie['Series_Title'] , " --- ", movie['IMDB_Rating'] )


# Main program
print("🎬 Welcome to Movie Recommendation Assistant 🎬")
name = input("\nWhat's your name? ")
print("\nHello ", name, "!")

# Select genre
genre = get_genre()

# Show recommendations
recommend(genre)
```
